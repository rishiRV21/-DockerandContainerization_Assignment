# Docker and Containerization Assignment

1. Install Docker and create Dockerfiles to containerize applications?

 -✅ Aim

To install Docker and create a Dockerfile to containerize a Python Flask web application.

✅ Tools/Software Used

Docker Desktop (Windows)

WSL Ubuntu

Docker Engine

Python Flask application

✅ Procedure
1. Verified Docker installation
docker --version
docker ps

2. Created project directory
mkdir docker-assignment
cd docker-assignment

3. Created Flask application file app.py
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return "Hello PW Skills! Docker container running successfully ✅"

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000)

4. Created dependency file requirements.txt
Flask==3.0.0

5. Created Dockerfile
FROM python:3.10-slim
WORKDIR /app
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt
COPY app.py .
EXPOSE 5000
CMD ["python", "app.py"]

6. Built Docker image
docker build -t pw-flask-app .

7. Ran container from image
docker run -d -p 5000:5000 --name pwcontainer pw-flask-app

8. Verified container running
docker ps

9. Verified application output
curl http://localhost:5000


Also verified in browser:

http://localhost:5000

✅ Result

Docker was successfully installed. A Dockerfile was created and the Flask application was containerized successfully. The container ran successfully and the application was accessed on localhost.

✅ Conclusion

Docker containerization was completed successfully using Dockerfile. The application was packaged into an image and executed inside a container.
